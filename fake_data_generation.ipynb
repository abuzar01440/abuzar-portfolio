```python
import pandas as pd
from faker import Faker
import random
from datetime import datetime, timedelta

# Initialize the Faker library to generate fake data
fake = Faker()

# --- 1. Generate Departments Data ---
print("Generating departments data...")
departments = ['Human Resources', 'Engineering', 'Marketing', 'Sales', 'IT', 'Finance']
department_data = {
    'department_id': range(1, len(departments) + 1),
    'department_name': departments
}
departments_df = pd.DataFrame(department_data)
print("Departments data generated.")

# --- 2. Generate Employees Data ---
print("Generating employees data...")
num_employees = 99
employees_list = []
for i in range(1, num_employees + 1):
    first_name = fake.first_name()
    last_name = fake.last_name()
    employees_list.append({
        'employee_id': 1000 + i,
        'first_name': first_name,
        'last_name': last_name,
        'email': f"{first_name.lower()}.{last_name.lower()}@{fake.free_email_domain()}",
        'hire_date': fake.date_between(start_date='-5y', end_date='today'),
        # Assign a random department to each employee
        'department_id': random.choice(departments_df['department_id'])
    })

employees_df = pd.DataFrame(employees_list)
print("Employees data generated.")


# --- 3. Generate Salaries Data ---
print("Generating salaries data...")
salaries_list = []
salary_id_counter = 1
for emp_id in employees_df['employee_id']:
    # Each employee can have one or more salary entries
    num_salary_entries = random.randint(1, 2)
    start_date = employees_df[employees_df['employee_id'] == emp_id]['hire_date'].iloc[0]

    for _ in range(num_salary_entries):
        from_date = start_date
        to_date = from_date + timedelta(days=random.randint(365, 730))

        salaries_list.append({
            'salary_id': salary_id_counter,
            'employee_id': emp_id,
            'salary': round(random.uniform(45000.00, 120000.00), 2),
            'from_date': from_date,
            'to_date': to_date
        })
        start_date = to_date + timedelta(days=1)
        salary_id_counter += 1

salaries_df = pd.DataFrame(salaries_list)
# Ensure we don't exceed the 99 row limit for this sheet either
if len(salaries_df) > 99:
    salaries_df = salaries_df.head(99)
print("Salaries data generated.")


output_filename = 'company_data.xlsx'
print(f"Writing data to {output_filename}...")
with pd.ExcelWriter(output_filename, engine='openpyxl') as writer:
    departments_df.to_excel(writer, sheet_name='departments', index=False)
    employees_df.to_excel(writer, sheet_name='employees', index=False)
    salaries_df.to_excel(writer, sheet_name='salaries', index=False)

print("\nDone! Your data has been generated and saved to 'company_data.xlsx'.")
```
