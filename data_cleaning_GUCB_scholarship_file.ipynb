```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("/content/planilha_ies.xlsx")
df.head(12)
```

```python
# remove the first five rows
df = df.iloc[5:]
df.head()
```

```python
#remove the first three columns
df = df.drop(df.columns[:3], axis=1)
df.head()
```

```python
# remove the 2md row
df = df.drop(df.index[1])
df.head()
```

```python
#remove the header and make the row 1 header and reset the index
df = df.iloc[1:]
df.head()
```

```python
df.columns = df.iloc[0]
df = df[1:].reset_index(drop=True)
display(df.head())
```

```python
df.head(2)
```

```python
df.columns
```

```python
df.Mestrado.unique()
```

```python
df = df[df['Mestrado'] != 'Mestrado'].reset_index(drop=True)
print(f"Unique values in Mestrado column: {df['Mestrado'].unique()}")
display(df.head())
```

```python
#remove the rows if they nan value in Mestrado column
df = df.dropna(subset=['Mestrado'])
print(f"Unique values in Mestrado column: {df['Mestrado'].unique()}")
display(df.head())
```

```python
df.Mestrado.value_counts()
```

```python
df.columns
```

```python
df = df.loc[:, df.columns.notna()]
display(df.head())
```

```python
df.describe()
```

```python
print(f'Total rows: {df.shape[0]}')
print(f'Total columns: {df.shape[1]}')
df.shape
```

```python
df.to_excel('cleaned_data.xlsx', index=False)
print('File saved as cleaned_data.xlsx')
```
