```python

# Convert date columns to datetime objects
date_cols = ['date_activ', 'date_end', 'date_modif_prod', 'date_renewal']
for col in date_cols:
    df_client[col] = pd.to_datetime(df_client[col])

# 1.  Churn Analysis

#   * Calculate the overall churn rate.
churn_rate = df_client['churn'].mean()
print(f"Overall Churn Rate: {churn_rate:.2%}")

#   * Churn rate by categorical variables (e.g., 'channel_sales', 'has_gas', 'origin_up').
categorical_cols = ['channel_sales', 'has_gas', 'origin_up']
for col in categorical_cols:
    churn_by_category = df_client.groupby(col)['churn'].mean().reset_index()
    print(f"\nChurn Rate by {col}:")
    print(churn_by_category)

    plt.figure(figsize=(8, 6))
    sns.barplot(x=col, y='churn', data=churn_by_category)
    plt.ylabel('Churn Rate')
    plt.title(f'Churn Rate by {col}')
    plt.show()

# 2.  Customer Lifetime Analysis

#   * Calculate customer lifetime (in years) as the difference between date_end and date_activ
df_client['customer_lifetime'] = (df_client['date_end'] - df_client['date_activ']).dt.days / 365.25

#   * Analyze the distribution of customer lifetimes.
plt.figure(figsize=(8, 6))
sns.histplot(df_client['customer_lifetime'], bins=20, kde=True)
plt.xlabel('Customer Lifetime (Years)')
plt.ylabel('Frequency')
plt.title('Distribution of Customer Lifetime')
plt.show()

#   * Explore the relationship between customer lifetime and churn.
plt.figure(figsize=(8, 6))
sns.boxplot(x='churn', y='customer_lifetime', data=df_client)
plt.xlabel('Churn')
plt.ylabel('Customer Lifetime (Years)')
plt.title('Customer Lifetime vs. Churn')
plt.show()


# 3.  Consumption and Forecasting Analysis

#   * Compare 'cons_12m' and 'forecast_cons_12m'.  Calculate the difference and percentage difference.
df_client['consumption_difference'] = df_client['forecast_cons_12m'] - df_client['cons_12m']
df_client['consumption_percentage_difference'] = (df_client['consumption_difference'] / df_client['cons_12m']) * 100

#   * Visualize the relationship between actual and forecasted consumption.
plt.figure(figsize=(8, 6))
sns.scatterplot(x='cons_12m', y='forecast_cons_12m', data=df_client)
plt.xlabel('Actual Consumption (12 months)')
plt.ylabel('Forecasted Consumption (12 months)')
plt.title('Actual vs. Forecasted Consumption')
plt.show()

#   * Analyze the distribution of the difference and percentage difference.
plt.figure(figsize=(8, 6))
sns.histplot(df_client['consumption_percentage_difference'], bins=20, kde=True)
plt.xlabel('Consumption Percentage Difference')
plt.ylabel('Frequency')
plt.title('Distribution of Consumption Forecast Accuracy')
plt.show()



# 4.  Price Sensitivity Analysis

#   * Explore the relationships between consumption and price-related features.
price_cols = ['forecast_price_energy_off_peak', 'forecast_price_energy_peak', 'forecast_price_pow_off_peak']
for col in price_cols:
    plt.figure(figsize=(8, 6))
    sns.scatterplot(x=col, y='cons_12m', data=df_client)
    plt.xlabel(col)
    plt.ylabel('Consumption (12 months)')
    plt.title(f'Consumption vs. {col}')
    plt.show()



#   * Analyze the distributions of 'margin_gross_pow_ele' and 'margin_net_pow_ele'.
margin_cols = ['margin_gross_pow_ele', 'margin_net_pow_ele']
for col in margin_cols:
    plt.figure(figsize=(8, 6))
    sns.histplot(df_client[col], bins=20, kde=True)
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.title(f'Distribution of {col}')
    plt.show()

#   * Explore the relationship between margins and churn.
for col in margin_cols:
    plt.figure(figsize=(8, 6))
    sns.boxplot(x='churn', y=col, data=df_client)
    plt.xlabel('Churn')
    plt.ylabel(col)
    plt.title(f'{col} vs. Churn')
    plt.show()
```

```python
df_price['price_date'] = pd.to_datetime(df_price['price_date'])

#   Explore the relationships between consumption and price-related features.
price_cols_var = ['price_off_peak_var', 'price_peak_var', 'price_mid_peak_var']
price_cols_fix = ['price_off_peak_fix', 'price_peak_fix', 'price_mid_peak_fix']

for col in price_cols_var:
    plt.figure(figsize=(8, 6))
    sns.scatterplot(x=col, y='cons_12m', data=df_client.merge(df_price, on='id', how='left')) #merge
    plt.xlabel(col)
    plt.ylabel('Consumption (12 months)')
    plt.title(f'Consumption vs. {col}')
    plt.show()

for col in price_cols_fix:
    plt.figure(figsize=(8, 6))
    sns.scatterplot(x=col, y='cons_12m', data=df_client.merge(df_price, on='id', how='left')) #merge
    plt.xlabel(col)
    plt.ylabel('Consumption (12 months)')
    plt.title(f'Consumption vs. {col}')
    plt.show()


#   * Analyze how prices have changed over time.
    # Group by 'price_date' and calculate the mean of price columns
df_price_mean = df_price.groupby('price_date')[price_cols_var + price_cols_fix].mean().reset_index()
print("_______________________________________________________________________")
print("Price mean")
print("_______________________________________________________________________")

print(df_price_mean)
print("_______________________________________________________________________")
#   * Analyze the distributions of 'margin_gross_pow_ele' and 'margin_net_pow_ele'.
margin_cols = ['margin_gross_pow_ele', 'margin_net_pow_ele']
for col in margin_cols:
    plt.figure(figsize=(8, 6))
    sns.histplot(df_client[col], bins=20, kde=True)
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.title(f'Distribution of {col}')
    plt.show()

#   * Explore the relationship between margins and churn.
for col in margin_cols:
    plt.figure(figsize=(8, 6))
    sns.boxplot(x='churn', y=col, data=df_client)
    plt.xlabel('Churn')
    plt.ylabel(col)
    plt.title(f'{col} vs. Churn')
    plt.show()
```
